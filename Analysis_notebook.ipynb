```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
country_list = list(gdp_raw.COUNTRY.unique())
country_gdp_dict = {}

for i in country_list:
    
    country_gdp_dict[f'{i}'] = (gdp_raw.loc[lambda x: (x['S_ADJUSTMENT'] == 'SA') & (x['TYPE_OF_TRANSFORMATION'] == 'XDC') & (x['COUNTRY'] == i) & (x['KEYWORDS[]'] == 'Real GDP')]
                                [['COUNTRY','TIME_PERIOD','OBS_VALUE']].assign(DATE = lambda x: pd.to_datetime(x['TIME_PERIOD']),
                                                                               GDP = lambda x: x["OBS_VALUE"])).set_index('DATE').drop(columns = ['TIME_PERIOD', 'OBS_VALUE'])
    

#gdp_filtered = gdp_raw.loc[lambda x: (x['S_ADJUSTMENT'] == 'SA') & (x['TYPE_OF_TRANSFORMATION'] == 'XDC')][['COUNTRY','TIME_PERIOD','OBS_VALUE']]
#gdp_filtered

country_gdp_dict['BRA']
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

bra_df = country_gdp_dict['BRA']
usa_df = country_gdp_dict['USA']

fig, ax = plt.subplots(1,2,figsize= (10, 5))
ax[0].set_title('This has to be linear otsebe to forecast (Brazil)')
ax[0].set_ylabel('GDP')
ax[0].set_xlabel("Time, Its quaterly hence why its linear")
sns.lineplot(data= bra_df, x= bra_df.index, y= 'GDP', ax = ax[0] )
ax[1].set_title('This has to be linear otsebe to forecast')
ax[1].set_ylabel('GDP')
ax[1].set_xlabel("Time, Its quaterly hence why its linear (USA)")
sns.lineplot(data= usa_df, x= bra_df.index, y= 'GDP', ax = ax[1] )

plt.show()
```
